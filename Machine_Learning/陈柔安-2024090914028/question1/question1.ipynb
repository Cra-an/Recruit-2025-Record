```python
import numpy as np
import matplotlib.pyplot as plt

def func(x):
    return 2 * x**5 + x**4 - 11 * x**3 + 7 * x**2 + 2 * x

def func_derivative(x):
    return 10 * x**4 + 4 * x**3 - 33 * x**2 + 14 * x + 2

learning_rate = 0.01
num_iterations = 500

x_current = np.random.uniform(-0.5, 1.5)
x_history = [x_current]
y_history = [func(x_current)]

for i in range(num_iterations):
    derivative = func_derivative(x_current)
    x_current = x_current - learning_rate * derivative
    x_current = np.clip(x_current, -0.5, 1.5)
    x_history.append(x_current)
    y_history.append(func(x_current))

x = np.linspace(-0.5, 1.5, 1000)
y = func(x)

plt.figure(figsize=(12, 8))
plt.plot(x, y, label='Function')
plt.plot(x_history, y_history, 'r-o', label='Gradient Descent Path')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Gradient Descent Process')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x_sample = np.linspace(-0.5, 1.5, 100)
y_sample = 2 * x_sample**5 + x_sample**4 - 11 * x_sample**3 + 7 * x_sample**2 + 2 * x_sample + np.random.normal(0, 0.1, size=100)

def predict(x, params):
    a, b, c, d, e = params
    return a * x**5 + b * x**4 + c * x**3 + d * x**2 + e * x

def loss(params):
    y_pred = predict(x_sample, params)
    return np.mean((y_sample - y_pred)**2)

def loss_gradient(params):
    a, b, c, d, e = params
    y_pred = predict(x_sample, params)
    error = y_sample - y_pred
    grad_a = -2 * np.mean(error * x_sample**5)
    grad_b = -2 * np.mean(error * x_sample**4)
    grad_c = -2 * np.mean(error * x_sample**3)
    grad_d = -2 * np.mean(error * x_sample**2)
    grad_e = -2 * np.mean(error * x_sample)
    return np.array([grad_a, grad_b, grad_c, grad_d, grad_e])

learning_rate = 0.01
num_iterations = 1000
params = np.random.rand(5)
loss_history = []

for i in range(num_iterations):
    gradient = loss_gradient(params)
    params = params - learning_rate * gradient
    current_loss = loss(params)
    loss_history.append(current_loss)

print(f"拟合得到的系数: {params}")

plt.figure(figsize=(10, 6))
plt.plot(loss_history)
plt.xlabel('Iteration')
plt.ylabel('Loss')
plt.title('Loss Function During Gradient Descent')
plt.grid(True)
plt.show()
```
